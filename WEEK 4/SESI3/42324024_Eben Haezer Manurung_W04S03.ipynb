# PRAKTIKUM

In [1] Import library dan memuat Iris. Library dasar di-import dan dataset Iris dimuat dari Scikit-Learn.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

# Load dataset Iris dari scikit-learn
iris = load_iris()
```

In [2] Konversi ke DataFrame. Data diubah menjadi DataFrame. X berisi 4 fitur numerik (sepal length, sepal width, petal length, petal width), y berisi nama spesies.

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["species"] = iris.target_names[iris.target]

# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["species"])
y = df["species"]

print(X)
```

In [3] Melihat target (y). Kolom species ditampilkan, berisi 150 data dengan tiga kelas: setosa, versicolor, virginica.

```python
y
```

In [4] Encode label. LabelEncoder mengubah nama spesies menjadi angka: setosa = 0, versicolor = 1, virginica = 2, agar bisa diproses model.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

In [5] Pembagian data. Data dibagi 80% latih (120 data) dan 20% uji (30 data) dengan train_test_split.

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

In [6] Melatih model Gaussian Naïve Bayes. Karena semua fitur Iris bertipe numerik kontinu, digunakan Gaussian Naïve Bayes, yang mengasumsikan tiap fitur mengikuti distribusi normal pada setiap kelas. Model dilatih dengan fit(X_train, y_train).

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

In [7] Prediksi data uji. Model memprediksi kelas untuk 30 data uji.

```python
result = NB_model.predict(X_test)
result
```

In [8] Evaluasi akurasi. Akurasi = 96,67% (29 dari 30 data uji benar), menunjukkan Gaussian Naïve Bayes bekerja sangat baik pada dataset Iris.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

In [9] Prediksi input baru (versi awal). Input [[5.1, 3.5, 1.4, 0.2]] diprediksi sebagai setosa. Namun muncul UserWarning karena input berupa list/array biasa tanpa nama kolom, sedangkan model dilatih dengan DataFrame yang punya nama kolom (iris.feature_names).

```python
#prediksi inputan data baru
#input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

In [10] Prediksi input baru (versi diperbaiki). Input diubah menjadi DataFrame dengan nama kolom yang sama seperti data latih, sehingga peringatan hilang. Hasilnya tetap setosa, karena hanya struktur datanya yang diperbaiki, bukan nilainya.

```python
input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```
