```python
import pandas as pd
df = pd.read_csv('Salary_Data.csv')
print(df.head())
```

```python
X = df.iloc[:, :- 1].values
y = df.iloc[:, 1].values
```

```python
import matplotlib.pyplot as plt
plt.scatter(X, y, color = "red")
plt.xlabel('Years Experience')
plt.ylabel('Salary')
plt.show()
```

```python
#bagi dataset jadi train dan test 80:20
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

```python
# Fitting Simple Linear Regression to the Training set
from sklearn. linear_model import LinearRegression
regressor = LinearRegression()
regressor .fit(X_train, y_train)
```

```python
a = regressor.intercept_
b = regressor.coef_[0]

print("Estimated model intercept, a:", a)
print("Estimated model slope, b:", b)
print(f"ŷ = {a:.2f} + {b:.2f}x")
```

```python
# memprediksi hasil pada data uji (Test set) menggunakan model regresi yang tela
y_pred = regressor.predict(X_test)
y_pred
```

```python
# Menghitung nilai R-Square, yang mengukur seberapa baik model regresi linear
# Semakin mendekati 1, semakin baik model

from sklearn.metrics import r2_score
score=r2_score(y_test,y_pred)
score
```

```python
# Visualizing the Training set results
plt.scatter(X_train, y_train, color = 'red')
plt.plot(X_train, regressor.predict(X_train), color = 'blue')
```

```python
# Visualizing the Test set results
plt.scatter(X_test, y_test, color = 'red')
plt.plot(X_train, regressor.predict(X_train), color = 'blue')
plt.title('Salary vs Experience (Test set)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

## Tugas 1: Simple Linear Regression (Student Score)

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

df = pd.read_csv('rounded_hours_student_scores.csv')
print(df.head())
print(df.isnull().sum())
```

```python
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
```

```python
plt.scatter(X, y, color='red')
plt.xlabel('Waktu Belajar (jam)')
plt.ylabel('Score')
plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

```python
a = regressor.intercept_
b = regressor.coef_[0]

print('Estimated model intercept, a:', a)
print('Estimated model slope, b:', b)
print(f'ŷ = {a:.2f} + {b:.2f}x')
```

```python
y_pred = regressor.predict(X_test)
hasil = pd.DataFrame({'Waktu Belajar': X_test.flatten(), 'Aktual': y_test, 'Prediksi': y_pred})
print(hasil)
```

```python
y_pred = regressor.predict(X_test)
hasil = pd.DataFrame({'Waktu Belajar': X_test.flatten(), 'Aktual': y_test, 'Prediksi': y_pred})
print(hasil)
```

```python
score = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
print('R-Square:', score)
print('MAE    :', mae)
print('RMSE   :', rmse)
```

```python
plt.scatter(X_train, y_train, color='red')
plt.plot(X_train, regressor.predict(X_train), color='blue')
plt.title('Score vs Waktu Belajar (Training set)')
plt.xlabel('Waktu Belajar (jam)')
plt.ylabel('Score')
plt.show()
```

```python
plt.scatter(X_test, y_test, color='red')
plt.plot(X_train, regressor.predict(X_train), color='blue')
plt.title('Score vs Waktu Belajar (Test set)')
plt.xlabel('Waktu Belajar (jam)')
plt.ylabel('Score')
plt.show()
```

```python
print('Prediksi score untuk 7.5 jam belajar:', regressor.predict([[7.5]])[0])
```

## Tugas 2: Multiple Linear Regression (Marketing)

```python
df2 = pd.read_csv('Marketing_Data.csv')
df2.columns = df2.columns.str.strip().str.lower()
print(df2.head())
print(df2.isnull().sum())
```

```python
print(df2.corr())
```

```python
X2 = df2[['youtube', 'facebook', 'newspaper']]
y2 = df2['sales']
```

```python
X_train2, X_test2, y_train2, y_test2 = train_test_split(X2, y2, test_size=0.2, random_state=0)
```

```python
regressor2 = LinearRegression()
regressor2.fit(X_train2, y_train2)
```

```python
a2 = regressor2.intercept_
b1, b2, b3 = regressor2.coef_

print('Estimated model intercept, a:', a2)
print('Estimated model slope, b1 (youtube):', b1)
print('Estimated model slope, b2 (facebook):', b2)
print('Estimated model slope, b3 (newspaper):', b3)
print(f'ŷ = {a2:.2f} {b1:+.4f}x1 {b2:+.4f}x2 {b3:+.4f}x3')
```

```python
y_pred2 = regressor2.predict(X_test2)
hasil2 = pd.DataFrame({'Aktual': y_test2.values, 'Prediksi': y_pred2})
print(hasil2.head(10))
```

```python
score2 = r2_score(y_test2, y_pred2)
mae2 = mean_absolute_error(y_test2, y_pred2)
rmse2 = np.sqrt(mean_squared_error(y_test2, y_pred2))
print('R-Square:', score2)
print('MAE    :', mae2)
print('RMSE   :', rmse2)
```

```python
plt.scatter(y_test2, y_pred2, color='blue')
plt.plot([y_test2.min(), y_test2.max()], [y_test2.min(), y_test2.max()], color='red')
plt.xlabel('Sales Aktual')
plt.ylabel('Sales Prediksi')
plt.title('Aktual vs Prediksi (Test set)')
plt.show()
```

```python
koef = pd.Series(regressor2.coef_, index=X2.columns)
koef.plot(kind='bar', color='teal')
plt.title('Koefisien Regresi per Media')
plt.ylabel('Koefisien')
plt.show()
```
